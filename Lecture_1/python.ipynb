## Some quick recaps about Python 
And some common mistakes that even very experienced programmers can get tripped over.

### Variable, operations

```python
# Give a value to a variable
x = 3
print(x)
```

```python
# Priority of operations
print(3 + 4 * 5)
print((3 + 4) * 5)
print(100 / 2 ** 2)
print((100 / 2) ** 2)
```

### List

```python
# Create and modify values of a list
mylist = [1,2,3]
print(mylist)
mylist[0] = 's'  # List counts from zero
print(mylist)
```

```python
# Slicing a list
mylist = [1,2,3]
print(mylist)
print(mylist[0:2])  # Includes the start, but not the end
```

```python
# Append vs insert
mylist = [1,2,3]
mylist.append('a')
print(mylist)
mylist.insert(1, 'd')
print(mylist)
```

```python
# Pop is the opposite of insert
mylist = [1,2,3]
mylist.pop(1)  # This returns the item pop'ed from the list
print(mylist)
```

### Dictionary

```python
# Dictionary creation
my_dictionary = {'a': [1,2,3], 'b': [4,5,6], 'c': [7,8,9]}
print(my_dictionary)
```

```python
# Visit item by key.
print(my_dictionary['a'])

# Update an item
my_dictionary['a'] = [-1, -1, -1]
print(my_dictionary['a'])

# Add an item to a dictionary
my_dictionary['d'] = ['asdfg']
print(my_dictionary['d'])
```

### Function

```python
# Function
y = -1
def some_function(x):
    y = x + 1  # When creating a variable within a function, Python gives a local copy. We'll revisit what "copy" means.
    return y 
print(some_function(2))
```

```python
# The local values are encapsulated within the scope of the function
print(y)
```

```python
# You can modify the list elements in-place.
def modify_list(L):
    L[0] = 'a' 

L = [1,2,3]
modify_list(L)
print(L)
```

```python
# Revisiting that "local copy"
def modify_list_2(L):
    L = ['a', 's', 'd']  # This sentence gives the local variable L the address of the list ['a', 's', 'd']
    # And the local variable L is a local copy.

L = [1,2,3]
modify_list_2(L)
print(L)
```

```python
# Creating a stateless function with lambda
# Aside: lambda comes from Lambda calculus. Many programming languages support this operation
f = lambda x: [x+1, x+2, x+3]
print(f(1))
print(f(2))
```

### Loops

```python
# "For" loop
mylist = ['s', 1, 2, 3]
for i in mylist:
    print(i)
```

```python
# "While" loop
i = 0
while i < len(mylist):
    print(mylist[i])
    i += 1  # Commenting out this line will cause infinite loop. Guess why?
```

```python
# range
# range(start, stop, step) gives an iterator
print(range(0, 10, 2))
# You can convert the iterator into a list
print(list(range(0, 10, 2)))
```

### Conditions

```python
def how_is_my_score(score):
    if score > 90:
        return "Good"
    elif score > 80:
        return "Ok"
    else:
        return "Needs improvement"

how_is_my_score(85)
```

```python
def how_is_my_score_2(score):
    dad_says, mom_says = "", ""
    if score > 80:
        dad_says = "Good"
    if score > 90:
        mom_says = "Good"
    elif score > 80:
        mom_says = "Ok"
    else:
        mom_says = "Needs improvement"
    return dad_says, mom_says 

print(how_is_my_score_2(85))
# print(how_is_my_score_2(75))  # How about this one?
```

### Error handling

```python
# Throwing (raising) an error
def divide(x, y):
    if y == 0:
        raise ValueError("The dividend shouldn't be zero!")
    else:
        return x / y 

print(divide(3, 2))
print(divide(4, 0))
```

```python
# Catching an error
try:
    divide(4, 0)
except ValueError:
    print("Caught this error!")
```

### Object
1. The "self" argument  
2. The `__len__` method.  
3. Instantiate an object.  

```python
class DataLoader:
    def __init__(self, data):
        self.data = data 

    def __len__(self):
        print("This is where the __len__() is called")
        return len(self.data)

    def one_batch(self, start_id, batch_size):
        return self.data[start_id : start_id + batch_size]

dl = DataLoader(list(range(8)))  # dl is an instance of the class DataLoader
print(len(dl))
```

```python
print(dl.one_batch(start_id=3, batch_size=4))
```

```python
# Change the one_batch() method into __next__, so you can use Python's built-in next() method.
class BetterDataLoader:
    def __init__(self, data, batch_size):
        self.data = data 
        self.i = 0
        self.batch_size = batch_size

    def __len__(self):
        return len(self.data)

    def __next__(self):
        if self.i + self.batch_size > len(self.data):
            raise StopIteration 
        batch = self.data[self.i : self.i + self.batch_size]
        self.i = self.i + self.batch_size
        return batch 

dl = BetterDataLoader(list(range(10)), 4)
print(next(dl))
print(next(dl))
print(next(dl))  # Throws a "StopIteration" error
```

```python
# Can we use for loop? We need to make the dataloader an **Iterable**. 
# i.e., give it a __iter__() method.

class IterableDataLoader:
    def __init__(self, data, batch_size):
        self.data = data 
        self.i = 0
        self.batch_size = batch_size

    def __len__(self):
        return len(self.data)

    def __next__(self):
        if self.i + self.batch_size > len(self.data):
            raise StopIteration 
        batch = self.data[self.i : self.i + self.batch_size]
        self.i = self.i + self.batch_size
        return batch 

    def __iter__(self):
        return self

dl = IterableDataLoader(list(range(10)), 5)
for batch in dl:
    print(batch)
```

```python
# We can also use pytorch's dataloader
# Note: we'll see more about pytorch in another tutorial. 
# For now let's just take the DataLoader as a tool
# More details in https://pytorch.org/docs/stable/data.html

import torch 

def my_collate_function(batch):
    """
    A collate function following the format requirement by pytorch dataloader.
    Input: batch, which essentially is a list of items
    """
    # result = []
    # for item in batch:
    #     result.append(item)
    # return torch.tensor(result)
    return torch.tensor(batch)

dl = torch.utils.data.DataLoader(
    dataset=list(range(10)),  # Here we define the item in the dataset
    batch_size=4,
    collate_fn=my_collate_function,
    drop_last=True
)
for batch in dl:
    print(batch)
```

### Saving and loading objects (using pickle)
You can use pytorch later.

```python
import pickle 

dl = IterableDataLoader(list(range(10)), 5)
with open("tmp.pkl", "wb") as f:  # Note the "wb" option -- it means binary write operation
    pickle.dump(dl, f)
```

```python
with open("tmp.pkl", "rb") as f:  # "rb" means binary read operation
    new_dl = pickle.load(f)

for batch in new_dl:
    print(batch)
```
