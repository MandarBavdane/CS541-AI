## Numpy

```python
import numpy as np 
```

### Numpy vectorizes the operations
$$x = \begin{bmatrix} 1, 2, 3 \end{bmatrix}, y = \begin{bmatrix} 4, 5, 6 \end{bmatrix}, x+y=?$$

```python
x = np.array([1,2,3])
y = np.array([4,5,6])
x + y 
```

```python
# Numpy array vs n-dimensional list

x = np.array([1,2,3])
print(x.shape)

x = np.array([[1,2,3], [4,5,6]])
print(x.shape)
```

```python
# Transpose
x = np.array([[1,2,3], [4,5,6]])
x, x.T 
```

```python
# reshape
x.reshape((1,6)), x.reshape((6,1)), x.reshape((2,3))
```

```python
# zeros_like and ones_like
z = np.zeros_like(x)
z
```

```python
# Numpy array <-> python list
z.tolist()
```

```python
# Mean, variance, std
x.mean(), x.var(), x.std()
```

```python
# Mean by dimension / axis
x.mean(axis=0), x.mean(axis=1)
```

```python
# Broadcasting
print(x)
x + np.array([1,2,3])
```

```python
x + np.array([[1,2,3], [1,2,3]])
```
