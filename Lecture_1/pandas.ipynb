### Create dataframe

```python
df = pd.DataFrame(data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
})
df
```

### Editing the data in a Dataframe
If you want to edit the data, use either `loc` or `iloc`

```python
print(df)
df.loc[0, "City"] = 'somewhere'  # access both the row and column via the name (here the row index happens to by numeric)
print(df)
```

```python
print(df)
df.iloc[0, 2] = 'nowhere'  # Need to access both row and column via the index
print(df)
```

```python
# While iterating through the dataframe, you'll see a copy.
for i, row in df.iterrows():
    #print(i)
    #print(row)
    if i == 0:
        row['City'] = -1  # Editing the copy doesn't affect the dataset itself
df 
```

### Aggregation
Two types of merging operations are commonly used: horizontal (`pd.merge`), vertical (`pd.concat`). Let's look at an example each.

```python
df1 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David']
})

df2 = pd.DataFrame({
    'id': [3, 4, 5, 6],
    'age': [25, 30, 35, 40]
})

df3 = pd.DataFrame({
    'id': [7, 8],
    'name': ['Eve', 'Frank']
})

pd.merge(df1, df2, on='id', how='inner')
```

```python
pd.merge(df1, df2, on='id', how='outer')
```

```python
# pd.concat([df1, df3], ignore_index=True)
pd.concat([df1, df3], ignore_index=False)  # See how the df3's index is unchanged
```

```python
pd.concat([df1, df2], ignore_index=False)
```

### Cross product
Cross product is also implemented in pandas. When you do cross product of dataframes A and B, any of A's row is horizontally merged with any of B's row.

```python
df1.merge(df2, how='cross')
```
